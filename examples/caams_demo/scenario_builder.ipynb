```julia
function animate_full_solution(env; steps = 100)
    sim_time = sum(maneuver.length for maneuver in env.leader_path)
    modified_range = range(env.solution.t[1], stop=sim_time, length=steps)

    n_agents = length(env.agents)
    n_savevals = length(env.composites.saveval)

    function plot_leader_path_t(t)
        # Create a function to plot the leader path in 2D
        maneuver_lengths = [maneuver.length for maneuver in env.leader_path]
        cumulative_lengths = cumsum(maneuver_lengths)
        idx = findfirst(x -> x >= t, cumulative_lengths)

        if isnothing(idx)
            return Dubins3D.compute_at_len(env.leader_path[end], env.leader_path[end].length)[1:2]
        end

        maneuver = env.leader_path[idx]
        t_maneuver = t - (idx == 1 ? 0 : cumulative_lengths[idx-1])
        t_maneuver = clamp(t_maneuver, 0.0, maneuver.length)

        # Only take x, y for 2D
        pos = Dubins3D.compute_at_len(maneuver, t_maneuver)
        return pos[1:2]
    end

    function plot_agent_path_t(t, agent_nominal, agent_backup, switchtime)
        if t >= switchtime
            backup_t = t - switchtime
            maneuver_lengths = [maneuver.length for maneuver in agent_backup]
            cumulative_lengths = cumsum(maneuver_lengths)

            total_length = sum(maneuver_lengths)
            idx = findfirst(x -> x >= backup_t, cumulative_lengths)
            if isnothing(idx)
                # return [0.0, 0.0]
                return Dubins3D.compute_at_len(agent_backup[end], agent_backup[end].length)[1:2]
            end

            maneuver = agent_backup[idx]
            t_maneuver = backup_t - (idx == 1 ? 0 : cumulative_lengths[idx-1])
            t_maneuver = clamp(t_maneuver, 0.0, maneuver.length-0.01)

            return Dubins3D.compute_at_len(maneuver, t_maneuver)[1:2]
        else
            return agent_nominal(t)
        end
    end

    anim = @animate for (idx, t) in enumerate(modified_range)
        println("Animating step $idx of $steps at time $t")

        # Plot setup
        p = plot()

        xlims!(p, env.scenario.domain_min[1]-0.5, env.scenario.domain_max[1]+0.5)
        ylims!(p, env.scenario.domain_min[2]-0.5, env.scenario.domain_max[2]+0.5)

        # Plot the obstacles
        for obstacle in env.obstacles
            plot!(p, get_2d_repr(obstacle), label = false, linewidth=2,fill=true, alpha=0.5)
        end

         # Plot the full leader path
        plot!(
            τ -> plot_leader_path_t(τ)[1],
            τ -> plot_leader_path_t(τ)[2],
            0.0, sim_time,
            label = "Leader Path",
            linewidth=2,
            color=:red
        )


        state_matrix = env.solution(t)

        # Plot agents positions
        plot!(p, state_matrix[:, 1], state_matrix[:, 2], seriestype=:scatter, label=false, markersize=3)

        # Plot the inter-agent collision radius as circles around each agent
        δ = env.agent_settings.agent_radius / 2.0
        θ = range(0, 2π, length=100)
        for agent in eachrow(state_matrix)
            x = agent[1]
            y = agent[2]

            circle_x = x .+ δ .* cos.(θ)
            circle_y = y .+ δ .* sin.(θ)
            plot!(p, circle_x, circle_y, alpha=0.5, label=false)
        end

        # Find the index of the saveval to use
        # saveval_idx = findlast(x -> x.t < t, env.composites.saveval)
        saveval_idx = floor(n_savevals * (t / env.solution.t[end])) + 1
        saveval_idx = min(Int(saveval_idx), n_savevals)


        # Plot paths taken by all agents up to this point in time
        for agent in 1:n_agents
            plot!(
                p,
                τ -> env.solution(τ)[agent, 1],
                τ -> env.solution(τ)[agent, 2],
                0.0, min(t, sim_time),
                label = "Agent $agent Path",
                linewidth=1,
                color=:black,
                linestyle = :solid
            )
        end

        # For each agent, plot the nominal up to the switch time, then the backup
        for agent in 1:n_agents
            agent_state = state_matrix[agent, :]

            # Get the agent's composite
            agent_nominal = env.composites.saveval[saveval_idx].nominal[agent]
            agent_backup = env.composites.saveval[saveval_idx].backup[agent]
            agent_switchtime = env.composites.saveval[saveval_idx].switch_time[agent]

            agent_pathlen = agent_switchtime + sum(x -> x.length, agent_backup)

            plot!(
                τ -> plot_agent_path_t(τ, agent_nominal, agent_backup, agent_switchtime)[1],
                τ -> plot_agent_path_t(τ, agent_nominal, agent_backup, agent_switchtime)[2],
                t, min(t + agent_pathlen, sim_time),
                label = "Agent $agent Committed",
                linewidth=2,
                color=:blue,
                linestyle = :dash
            )
        end

        # for agent in eachindex(env.agents)
        #     subprob = GatekeeperFormationFlight.MultiGatekeeper.get_single_agent_subproblem(
        #         env.gk.problem,
        #         agent
        #     )

        #     plot!(
        #         τ -> GatekeeperFormationFlight.Gatekeeper.get_reference_state_and_input(
        #             subprob,
        #             subprob.reference_path,
        #             subprob.offset,
        #             τ
        #         )[1][1],
        #         τ -> GatekeeperFormationFlight.Gatekeeper.get_reference_state_and_input(
        #             subprob,
        #             subprob.reference_path,
        #             subprob.offset,
        #             τ
        #         )[1][2],
        #         0.0,
        #         sim_time,
        #         label = "Agent $agent Nominal",
        #         linewidth=2,
        #         color=:green,
        #         linestyle = :dot,
        #         alpha = 0.5
        #     )
        # end

        ## TODO Plot agent offset trajectories / nominals

    end

    return anim
end
```
